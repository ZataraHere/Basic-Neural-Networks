```python
import numpy as np
```

```python
a3 = np.random.rand(8,4)
a3
```

```python
keep_prob = 0.8
d3 = np.random.rand(a3.shape[0], a3.shape[1])<keep_prob
a3 = np.multiply(a3, d3)
# Inverted Dropout
a3 /= keep_prob
a3
```
