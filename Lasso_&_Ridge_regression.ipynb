# ***Ridge Regression***

```python
import numpy as np


class RidgeRegression:

    def __init__(
        self,
        learning_rate=0.01,
        lambda_=1.0,
        epochs=1000
    ):
        self.learning_rate = learning_rate
        self.lambda_ = lambda_
        self.epochs = epochs

        self.weights = None
        self.bias = None

        self.cost_history = []

    # ==================================================
    # Initialization
    # ==================================================

    def initialize_parameters(self, n_features):

        self.weights = np.zeros(
            (n_features, 1)
        )

        self.bias = 0.0

    # ==================================================
    # Prediction
    # ==================================================

    def predict(self, X):

        return X @ self.weights + self.bias

    # ==================================================
    # Cost
    # ==================================================

    def compute_cost(self, X, y):

        m = X.shape[0]

        predictions = self.predict(X)

        error = predictions - y

        mse = (1 / (2 * m)) * np.sum(
            error ** 2
        )

        # L2 penalty
        regularization = (
            self.lambda_ / (2 * m)
        ) * np.sum(
            self.weights ** 2
        )

        return mse + regularization

    # ==================================================
    # Training
    # ==================================================

    def fit(self, X, y):

        m, n_features = X.shape

        self.initialize_parameters(
            n_features
        )

        for epoch in range(self.epochs):

            # --------------------------
            # Forward
            # --------------------------

            predictions = self.predict(X)

            error = predictions - y

            # --------------------------
            # Gradients
            # --------------------------

            dw = (
                X.T @ error
            ) / m

            # Ridge gradient
            dw += (
                self.lambda_ / m
            ) * self.weights

            db = np.mean(error)

            # --------------------------
            # Update
            # --------------------------

            self.weights -= (
                self.learning_rate * dw
            )

            self.bias -= (
                self.learning_rate * db
            )

            # --------------------------
            # Cost
            # --------------------------

            cost = self.compute_cost(
                X,
                y
            )

            self.cost_history.append(cost)

            if epoch % 100 == 0:

                print(
                    f"Epoch {epoch}, "
                    f"Cost: {cost:.6f}"
                )

        return self

    # ==================================================
    # R2 Score
    # ==================================================

    def r2_score(self, X, y):

        predictions = self.predict(X)

        ss_res = np.sum(
            (y - predictions) ** 2
        )

        ss_tot = np.sum(
            (y - np.mean(y)) ** 2
        )

        return 1 - (
            ss_res / ss_tot
        )
```

```python
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6]
], dtype=float)

y = np.array([
    [2],
    [4],
    [6],
    [8],
    [10],
    [12]
], dtype=float)
```

```python
model = RidgeRegression(
    learning_rate=0.01,
    lambda_=0.1,
    epochs=2000
)

model.fit(X, y)
```

```python
predictions = model.predict(X)

print(predictions)
print("R²:", model.r2_score(X, y))
```

# ***Lasso Regression***

```python
class LassoRegression:

    def __init__(
        self,
        learning_rate=0.01,
        lambda_=1.0,
        epochs=1000
    ):
        self.learning_rate = learning_rate
        self.lambda_ = lambda_
        self.epochs = epochs

        self.weights = None
        self.bias = None

        self.cost_history = []

    # ==================================================
    # Initialization
    # ==================================================

    def initialize_parameters(self, n_features):

        self.weights = np.zeros(
            (n_features, 1)
        )

        self.bias = 0.0

    # ==================================================
    # Prediction
    # ==================================================

    def predict(self, X):

        return X @ self.weights + self.bias

    # ==================================================
    # Cost
    # ==================================================

    def compute_cost(self, X, y):

        m = X.shape[0]

        predictions = self.predict(X)

        error = predictions - y

        mse = (
            1 / (2 * m)
        ) * np.sum(
            error ** 2
        )

        # L1 penalty
        regularization = (
            self.lambda_ / m
        ) * np.sum(
            np.abs(self.weights)
        )

        return mse + regularization

    # ==================================================
    # Training
    # ==================================================

    def fit(self, X, y):

        m, n_features = X.shape

        self.initialize_parameters(
            n_features
        )

        for epoch in range(self.epochs):

            # --------------------------
            # Forward
            # --------------------------

            predictions = self.predict(X)

            error = predictions - y

            # --------------------------
            # Gradient
            # --------------------------

            dw = (
                X.T @ error
            ) / m

            # L1 gradient
            dw += (
                self.lambda_ / m
            ) * np.sign(
                self.weights
            )

            db = np.mean(error)

            # --------------------------
            # Update
            # --------------------------

            self.weights -= (
                self.learning_rate * dw
            )

            self.bias -= (
                self.learning_rate * db
            )

            # --------------------------
            # Cost
            # --------------------------

            cost = self.compute_cost(
                X,
                y
            )

            self.cost_history.append(cost)

            if epoch % 100 == 0:

                print(
                    f"Epoch {epoch}, "
                    f"Cost: {cost:.6f}"
                )

        return self

    # ==================================================
    # R2 Score
    # ==================================================

    def r2_score(self, X, y):

        predictions = self.predict(X)

        ss_res = np.sum(
            (y - predictions) ** 2
        )

        ss_tot = np.sum(
            (y - np.mean(y)) ** 2
        )

        return 1 - (
            ss_res / ss_tot
        )
```

```python
model = LassoRegression(
    learning_rate=0.01,
    lambda_=0.1,
    epochs=2000
)

model.fit(X, y)

predictions = model.predict(X)

print(predictions)

print(
    "R²:",
    model.r2_score(X, y)
)
```
